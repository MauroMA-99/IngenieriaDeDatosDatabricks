```sql
%python
dbutils.widgets.removeAll()
```

```sql
create widget text storageLocation default "abfss://lhcldata@adlsedu01.dfs.core.windows.net";
create widget text catalogo default "adbedu12";
```

### Creacion de catalog

```sql
CREATE CATALOG IF NOT EXISTS adbedu12;
```

### Creacion de Schemas

```sql
CREATE SCHEMA IF NOT EXISTS adbedu12.bronze;
CREATE SCHEMA IF NOT EXISTS adbedu12.silver;
CREATE SCHEMA IF NOT EXISTS adbedu12.golden;
CREATE SCHEMA IF NOT EXISTS adbedu12.exploratory;
```

```sql
use catalog ${catalogo}
```

### Creacion tablas Bronze

```sql
CREATE OR REPLACE TABLE adbedu12.bronze.CLIENTES (
  id INT, nombre STRING, correo STRING, pais STRING
)
USING DELTA
LOCATION '${storageLocation}/bronze/CLIENTES';

CREATE OR REPLACE TABLE adbedu12.bronze.VENTAS (
  id INT, id_cliente INT, fecha DATE, monto DOUBLE
)
USING DELTA
LOCATION '${storageLocation}/bronze/VENTAS';

CREATE OR REPLACE TABLE adbedu12.bronze.PRODUCTOS (
  id INT, nombre STRING, categoria STRING, precio DOUBLE
)
USING DELTA
LOCATION '${storageLocation}/bronze/PRODUCTOS';
```

### Creacion tablas Silver

```sql
CREATE OR REPLACE TABLE adbedu12.silver.VENTAS_LIMPIAS (
  id INT, id_cliente INT, fecha DATE, monto DOUBLE
)
USING DELTA
LOCATION '${storageLocation}/silver/VENTAS_LIMPIAS';

CREATE OR REPLACE TABLE adbedu12.silver.CLIENTES_VALIDOS (
  id INT, nombre STRING, correo STRING, pais STRING, activo BOOLEAN
)
USING DELTA
LOCATION '${storageLocation}/silver/CLIENTES_VALIDOS';

CREATE OR REPLACE TABLE adbedu12.silver.DETALLE_PRODUCTOS (
  id INT, nombre STRING, categoria STRING, precio DOUBLE, stock INT
)
USING DELTA
LOCATION '${storageLocation}/silver/DETALLE_PRODUCTOS';
```

### Creacion tablas Golden

```sql
CREATE OR REPLACE TABLE adbedu12.golden.KPIS_VENTAS (
  fecha DATE, total_ventas DOUBLE, promedio_ticket DOUBLE
)
USING DELTA
LOCATION '${storageLocation}/golden/KPIS_VENTAS';

CREATE OR REPLACE TABLE adbedu12.golden.SEGMENTOS_CLIENTES (
  id_cliente INT, segmento STRING, valor_cliente DOUBLE
)
USING DELTA
LOCATION '${storageLocation}/golden/SEGMENTOS_CLIENTES';
```

### Creacion tablas Exploratory

```sql
CREATE OR REPLACE TABLE adbedu12.exploratory.EXPERIMENTOS (
  id INT, descripcion STRING, resultado STRING
)
USING DELTA
LOCATION '${storageLocation}/exploratory/EXPERIMENTOS';

CREATE OR REPLACE TABLE adbedu12.exploratory.PRUEBAS_MODELOS_EXT (
  id INT, modelo STRING, precision DOUBLE, fecha DATE
)
USING DELTA
LOCATION '${storageLocation}/exploratory/PRUEBAS_MODELOS_EXT';
```
